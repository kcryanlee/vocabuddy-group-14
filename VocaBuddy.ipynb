```python
import random

vocab = {
    "apple": "蘋果",
    "book": "書",
    "computer": "電腦",
    "competition": "競賽",
    "program": "程式",
    "python": "蟒蛇",
    "integer":"整數"
}

print("----------------")
print("VocaBuddy 單字庫")
while True:
  print("----------------")
  print("1 顯示單字")
  print("2 顯示單字總數")
  print("3 單字測驗")
  print("4 隨機抽取單字")
  print("5 離開")
  op = input("請選擇功能: ")
  print("----------------")
  if op == "1":
    print("1 顯示單字")
    for word, meaning in vocab.items():
        print(word, ":", meaning)
  elif op == "2":
    print("總共"+str(len(vocab))+"個單字")
  elif op == "3":
      word, meaning = random.choice(list(vocab.items()))
      print("單字測驗")
      answer = input(f"請問「{word}」的中文意思是什麼？")
      if answer == meaning:
        print("答對了！")
      else:
        print(f"答錯了，正確答案是：{meaning}")
  elif op == "4":
    word, meaning = random.choice(list(vocab.items()))
    print("隨機抽取單字")
    print(f"英文: {word}")
    print(f"中文: {meaning}")
  elif op == '5':
    print("Bye Bye")
    break
  else:
    print("\n請輸入1-5")
```
